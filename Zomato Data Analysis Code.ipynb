### **# Importing Dataset using Python Library**

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
import warnings 
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv(r"Zomato_DS.csv")
data
```

### **# Columns in the Dataset.**

```python
data.columns
```

### **# Drop down unnesscessary columns**

```python
data = data.drop(['Unnamed: 0.1', 'Unnamed: 0', 'restaurant name', 'local address'], axis = 1)
data
```

### **# Check for Null values**

```python
# Check for null values in each column

null_values = data.isnull().sum()
print(null_values[null_values > 0])
```

### **# If found, then fill the missing values**

```python
# Fill the missing values of "rate (out of 5)" using mean

mean_rate = data['rate (out of 5)'].mean()
data['rate (out of 5)'].fillna(mean_rate, inplace = True)

# Fill the missing values of "avg cost (two people)" using median

median_cost = data['avg cost (two people)'].median()
data['avg cost (two people)'].fillna(median_cost, inplace = True)
```

### **# Now we will go further for Data Analysis**

**We will Analyze and Visualize the dataset through questions**

### **Q1. What are the most common restaurant types in the dataset?**

#### **# Bar Plot to Visualize.**

```python
# Count the frequency of each restaurant type

# Statistically 
restaurant_type_counts = data['restaurant type'].value_counts().reset_index()
restaurant_type_counts.columns = ['restaurant type', 'count']
print(restaurant_type_counts)

# Creating a bar plot 
plt.figure(figsize=(20,15))
sns.barplot(x='count', y='restaurant type', data=restaurant_type_counts, palette='viridis')
plt.xlabel("Count")
plt.ylabel("Restaurant Type")
plt.grid(axis = 'x', linestyle = '--', alpha = 1)
plt.show()
```

- **The most common Restaurant type in this dataset is 'Quick Bites', with 2840 occurences.**
- **And the second most common Restaurant type is 'Casual Dining' with 1634 occurences.**
- **This indicates that 'Quick Bites' are the predominant type of restaurant in the dataset.**

### **Q2. What is the overall distribution of ratings across all restaurants?**

#### **Bar Plot to Visualize.**

```python
# Get summary statistics for the 'rate' column
rating_summary = data['rate (out of 5)'].describe()

# Get the count of each ratings
rating_counts = data['rate (out of 5)'].value_counts().sort_index()

print(rating_summary)
print(rating_counts)

plt.figure(figsize=(15,12))
sns.barplot(x=rating_counts.index, y=rating_counts.values, palette='viridis')
plt.title("Count of each Rating")
plt.xlabel("Rate (out of 5)")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation = 90)
plt.tight_layout()
plt.show()
```

#### **# Histogram for the same for better understanding.**

```python
plt.figure(figsize = (10,8))
sns.histplot(data['rate (out of 5)'], bins = 5, kde = True, color = 'teal')
plt.title("Overall Distribution of Ratings across all Restaurants")
plt.xlabel("Ratings (out of 5)")
plt.ylabel("Number of Restaurants")
plt.show()
```

- **The overall distribution of ratings across all restaurants, we can see that most of the restaurants have ratings between 3.3 to 3.9.**
- **From this we can say that the customers appear to be reasonably satisfied with their experience**
- **However, there is an opportunity for restaurants to enhance their offerings and move into higher rating brackets by focusing on areas that could further delight and exceed customer expectations.**

### **Q4. What is the average cost for two people at different restaurants?**
   - Analyze the distribution and central tendency (mean, median) of `avg cost (two people)`.

#### **# Bar Plot to Visualize.**

```python
# Calculate mean and median of 'avg cost (two people)'
mean_cost = data['avg cost (two people)'].mean()
median_cost = data['avg cost (two people)'].median()

print(f"Mean cost for two people : Rs. {mean_cost:.2f}")
print(f"Median cost for the two people : Rs. {median_cost:.2f}")

# Visual Representation : Plot Bar plot of average cost
plt.figure(figsize=(12,10))
sns.barplot(x = 'restaurant type', y = 'avg cost (two people)',data = data, palette = "ocean")
plt.title("Barplot of Average Cost for two people")
plt.xlabel("Restaurant Type")
plt.ylabel("Average Cost for two people")
plt.xticks(rotation = 90)
plt.tight_layout()
plt.show()
```

- **The result of bar plot clearly depicts that the average cost for two people at each type of restaurant is between 300 to 600 Rs/-**

### **Q5. How does the availability of online ordering and table booking vary across different restaurant types?**
   - Investigate the distribution of `online_order` and `table booking` across `restaurant type`.

#### **# Pie Charts for Online Ordering and Table Booking.**

```python
# Convert 'online_order' and 'table booking' to binary (1 for 'Yes', 0 for 'No')
data['online_order'] = data['online_order'].map({'Yes': 1, 'No': 0})
data['table booking'] = data['table booking'].map({'Yes': 1, 'No': 0})

# Calculate the overall percentage for online ordering
online_order_percent = data['online_order'].mean() * 100
table_booking_percent = data['table booking'].mean() * 100

# Data for pie charts
online_order_data = [online_order_percent, 100 - online_order_percent]
table_booking_data = [table_booking_percent, 100 - table_booking_percent]

# Labels for the pie charts
labels = ['Available', 'Not Available']

# Creating pie chart for online ordering
plt.figure(figsize = (12, 10))

plt.subplot(1, 2, 1)
plt.pie(online_order_data, labels = labels, autopct = '%1.1f%%', colors = ['teal', 'red'])

plt.title('Percentage of Restaurants Offering online ordering')

# Creating pie chart for table booking 
plt.subplot(1, 2, 2)
plt.pie(table_booking_data, labels = labels, autopct = '%1.1f%%', colors = ['royalblue', 'red'])
plt.title('Percentage of Restaurants Offering Table booking')

plt.show()
```

- **The above Pie charts, Percentage of Restaurants Offering online ordering - 52.5% of Restaurants offer online ordering facilities.**
- **Where as on the other hand ; Percentage of Restaurants Offering Table booking - Only 10.5% of the Restaurants offer table booking facilities.**
- **In summary, Restaurants receive Online Orders the most from customers ordering from Zomato Application.**

### **Q6. Which areas have the highest concentration of highly-rated restaurants?**

#### **# Bar Chart for areas having highest concentration of highly-rated Restaurants.**

```python
# Create a DataFrame
df = pd.DataFrame(data)

# Define a threshold for highly-rated restaurants 
rating_threshold = 4.0

# Filter highly-rated restaurants
highly_rated_df = df[df['rate (out of 5)'] > rating_threshold]

# Group by area and count the number of highly-rated restaurants
area_counts = highly_rated_df['area'].value_counts()

# Convert to DataFrame for easier plotting
area_counts_df = area_counts.reset_index()
area_counts_df.columns = ['area', 'count']

# Plot
plt.figure(figsize = (10, 8))
sns.barplot(x = 'area', y = 'count', data = area_counts_df, palette = 'magma')
plt.title("Number of Highly-rated Restaurants by Area")
plt.xlabel('Area')
plt.ylabel('Number of Highly-rated Restaurants')
plt.xticks(rotation = 90)      # For better readability
plt.tight_layout()
plt.show()
```

- **The above plot shows that which areas have the highest number of highly-rated Restaurants(ratings > 4).**
-  **Hence, the area : Byresandra, Tavarekere, Madiwala have the highly-rated Restaurants.**

### **Q7. Which cuisines are most commonly associated with high-rated restaurants?**

#### **# Bar Chart for cuisines that are most commonly associated with high-rated restaurants.**

```python
df = pd.DataFrame(data)

# Filter high-rated restaurants
high_rated = df[df['rate (out of 5)'] >= 4.5]

# Count the frequency of each cuisine type among high-rated restaurants
cuisine_count = high_rated['cuisines type'].value_counts()

# Visualize the results
plt.figure(figsize = (20,20))
sns.barplot(x = cuisine_count.values, y = cuisine_count.index, palette = 'viridis')
plt.title("Most common Cuisines in High-Rated Restaurants")
plt.xlabel('Number of Restaurants')
plt.ylabel('Cuisine Type')
plt.show()
```

- **The above bar chart shows the most common cuisines among high-rated Restaurant.**
- **We can find that cuisines like South Indian, Cafe, Beverages, North Indian, Continental and American are commonly found in restaurants with high-ratings, indicating their popularity or the quality of offerings in those categories.**

### **Q8. How does the availability of online ordering and table booking impact customer ratings?**

#### **# Bar Plot for the availability of online ordering and table booking if they have any impact customer ratings.**

```python
df['table booking'] = df['table booking'].map({0 : 'Not Available', 1 : 'Available'})
#df['online_order'] = df['online_order'].map({0 : 'Not Available', 1 : 'Available'})
# Group by 'online_order' and 'table booking' and calculate the average rating
grouped_data = df.groupby(['online_order', 'table booking'])['rate (out of 5)'].mean().reset_index()

# Visualize the impact of Online ordering on ratings
plt.figure(figsize = (10, 8))
sns.barplot(x = 'online_order', y = 'rate (out of 5)', hue = 'table booking', data = grouped_data, palette = 'Blues')
plt.title('Impact of Online ordering and Table booking on Customer Ratings')
plt.xlabel('Online Ordering Available')
plt.ylabel('Average Rating')
plt.legend(title = 'Table Booking Available')
plt.show()

print(grouped_data)
```

- **In the above bar plot, the 0 and 1 on the x-axis represents whether the online ordering is available or not(0 : No, 1 : Yes)**
- **where as on the other hand light light blue color(Available) represents that table booking is available in the Restaurant and dark blue color(Not Available) represents that table booking isn't available.**
  
- **The combination of both services doesn't necessarily result in the highest possible rating, but it still maintains a high rating close to that of table booking alone.**
- **Overall, it seems that providing table booking services appears to be the most impactful on customer satisfaction, followed by the availability of online ordering.**

### **Q9. How does the presence of Online ordering affect the average rating and cost?**

#### **# Bar Plot**

```python
# Group by 'online_order' and calculate the mean for rating and avg cost
grouped_df = df.groupby('online_order').agg({'rate (out of 5)' : 'mean','avg cost (two people)' : 'mean'}).reset_index()

print(grouped_df)

# Create Fig and axis
fig, ax = plt.subplots(1, 2, figsize = (15, 10))

# Plot Avg Rating
sns.barplot(x = 'online_order', y = 'rate (out of 5)', data = grouped_df, ax = ax[0], palette = 'grey')
ax[0].set_title('Average Rate by Online Order')
ax[0].set_ylabel('Average Rate')
ax[0].set_xlabel('Online Order')
ax[0].set_xticks([0,1])
ax[0].set_xticklabels(['No', 'Yes'])

# Plot Avg Cost
sns.barplot(x = 'online_order', y = 'avg cost (two people)', data = grouped_df, ax = ax[1], palette = 'viridis')
ax[1].set_title('Average Cost by Online Order')
ax[1].set_ylabel('Average Cost')
ax[1].set_xlabel('Online Order')
ax[1].set_xticks([0, 1])
ax[1].set_xticklabels(['No', 'Yes'])

plt.tight_layout()
plt.show()
```

- **From the analysis and visualizations, it's clear that restaurants with online ordering tend to have slightly higher ratings and lower costs compared to those without online ordering.**
- **This could imply that online ordering might be associated with better customer satisfaction and/or lower operational costs.**


### **Q10. Do restaurants that offer table booking have different average ratings compared to those that don't?**

```python
#df['table booking'] = df['table booking'].map({0 : 'Not Available', 1 : 'Available'})
# Group by 'table booking' and calculate the mean rating and avg cost
grouped_df = df.groupby('table booking').agg({'rate (out of 5)' : 'mean', 'avg cost (two people)' : 'mean'}).reset_index()
print(grouped_df)
# Create a Fig and axis
fig, ax = plt.subplots(1, 2, figsize = (15, 10))

# Plot average rating
sns.barplot(x = 'table booking', y = 'rate (out of 5)', data = grouped_df, ax = ax[0], palette = 'Set1')
ax[0].set_title('Average Rating of Table Booking')
ax[0].set_xlabel('Table Booking')
ax[0].set_ylabel('Average Rating')
ax[0].set_xticks([0, 1])
ax[0].set_xticklabels(['No', 'Yes'])

# Plot average cost
sns.barplot(x = 'table booking', y = 'avg cost (two people)', data = grouped_df, ax = ax[1], palette = 'Set2')
ax[1].set_title('Average Cost of Table Booking')
ax[1].set_xlabel('Table Booking')
ax[1].set_ylabel('Average Cost')
ax[1].set_xticks([0, 1])
ax[1].set_xticklabels(['No', 'Yes'])

plt.tight_layout()
plt.show()
```

- **The first plot compares the average ratings between restaurants with and without table booking.**
- **The second plot compares the average costs between the two groups.**

- **Restaurants with Table Booking have an average rating of 4.04, which is higher than those without table booking(3.45).**
- **The average cost for restaurants with table booking is Rs. 1354.91, significantly higher compared to those without table booking (Rs. 443.75).**
